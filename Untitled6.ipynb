```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
search_tool = DuckDuckGoSearchRun()
```

```python
@tool
def get_queue_data(department: str) -> str:
    """
    Get live queue information for hospital departments
    """

    hospital_data = {
        "Orthopedic": {
            "patients": 42,
            "wait_time": 55,
            "crowd": "High"
        },

        "ENT": {
            "patients": 12,
            "wait_time": 18,
            "crowd": "Low"
        },

        "Cardiology": {
            "patients": 30,
            "wait_time": 45,
            "crowd": "Medium"
        }
    }

    if department in hospital_data:

        data = hospital_data[department]

        return f"""
        Department: {department}
        Patients in Queue: {data['patients']}
        Estimated Wait Time: {data['wait_time']} mins
        Crowd Level: {data['crowd']}
        """

    return "Department data not available."
```
